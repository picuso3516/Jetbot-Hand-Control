```python
import socket
from jetbot import Robot

robot = Robot()

# ── map command → การเคลื่อนที่ ──────────────────────────
SPEED = 0.4   # ปรับความเร็วได้ที่นี่ 0.0-1.0

def move(cmd):
    if cmd == "forward":
        robot.forward(SPEED)
    elif cmd == "backward":
        robot.backward(SPEED)
    elif cmd == "left":
        robot.left(SPEED)
    elif cmd == "right":
        robot.right(SPEED)
    elif cmd == "spin":
        robot.set_motors(SPEED, -SPEED)   # หมุนรอบตัวเอง
    elif cmd == "stop":
        robot.stop()
    else:
        robot.stop()   # command ไม่รู้จัก = หยุด

# ── เปิด Socket Server ───────────────────────────────────
HOST = "0.0.0.0"   # รับจากทุก IP
PORT = 9999

server = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
server.setsockopt(socket.SOL_SOCKET, socket.SO_REUSEADDR, 1)
server.bind((HOST, PORT))
server.listen(1)
print(f"JetBot listening on port {PORT}...")
print(f"IP: ", end="")

import os
os.system("hostname -I")

conn, addr = server.accept()
print(f"Connected from: {addr}")

# ── รับและประมวลผล command ───────────────────────────────
try:
    buffer = ""
    while True:
        data = conn.recv(1024).decode()
        if not data:
            print("Connection closed")
            break

        buffer += data
        # แยก command ด้วย newline
        while "\n" in buffer:
            cmd, buffer = buffer.split("\n", 1)
            cmd = cmd.strip()
            if cmd:
                print(f"CMD: {cmd}")
                move(cmd)

except Exception as e:
    print(f"Error: {e}")

finally:
    print("Stopping robot...")
    robot.stop()
    conn.close()
    server.close()
    print("Server closed")
```
